## Mise en service et tests du serveur MLflow


**Objectif**  
Mettre en service un serveur MLflow. Tester un jeu d'enregistrement dans différents gestionnaires de versions Python.


Afin de vérifier l'installation du serveur MLflow, le script teste l'enregistrement de métriques, de paramètres et de fichiers JSON dans un serveur MLflow.

Les environnements sont UV et Conda.

Les scripts de configuration et de tests sont disponibles dans le répertoire: /src/mlflow



```python
from pathlib import Path
import mlflow
import json
import subprocess

import logging


# configuration des paramètres de connexion et des répertoires de travail
MLFLOW_URL = "http://localhost:5000"
DATA_DIR = str(Path.home() / "data")
EXPERIMENT_NAME = "Projet 06 - OpenClassrooms v4"

# répertoire de l'entrée MLproject
cwd = Path.cwd()
init_mlflow_dir = str(cwd.parent.resolve()) + "/src/mlflow"
print("Répertoire de l'entrée MLproject:\n", init_mlflow_dir)

# Script python lancé comme processus dans l'environnement UV (pyproject.toml)
result = subprocess.run(
    ["uv", "run", "python", init_mlflow_dir + "/check_mlflow.py",
     "--mlflow_url", MLFLOW_URL,
     "--data_dir", DATA_DIR,
     "--experiment_name", EXPERIMENT_NAME],
    cwd=init_mlflow_dir,
    capture_output=True,
    text=True
)

display('--------result = subprocess.run-----------', result)

# script python lancé dans l'environnement MLproject (src/mlflow/conda.yaml)
logging.getLogger("mlflow").setLevel(logging.ERROR)
print('--------mlflow.run------------')

result_mlflow = mlflow.run(
                        uri=init_mlflow_dir,
                        entry_point="check",
                        parameters={
                            "mlflow_url": MLFLOW_URL,
                            "data_dir": DATA_DIR,
                            "experiment_name": EXPERIMENT_NAME
                            },
                        )

display('--------result = mlflow.run-----------', result_mlflow.get_status(), result_mlflow.run_id)

   

```
